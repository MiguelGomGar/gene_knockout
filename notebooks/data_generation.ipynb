Since each raw data file weighs over 4 GB, loading and sorting them consumes a lot of time, so we will do this process once. To optimize the data processing, the processed file (a sorted dictionary) will be saved, therefore the file will be loaded much faster on the following trials.

In order to obtain the sorted dictionary for each cell line, the `cell_line` variable must be changed accordingly.

```
# cell lines: A549, HT29, PC3
path_root = "data/"
cell_line = "HT29"

print(f"Loading dataset for cell line: {cell_line}")
dataset = load_dataset_from_torch_data(cell_line=cell_line, path_root=path_root)

print("Saving cached data...")
cache_path = os.path.join(path_root, f"cached_dataset_{cell_line}.pt")
torch.save(dataset, cache_path)

print(f"Cached data saved to {cache_path}")
```
